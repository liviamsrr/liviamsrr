# Ask
Como achar o email institucional a partir do nome do aluno?

#Get the data



```python
from google.colab import files
uploaded = files.upload()
import io
data = io.BytesIO(uploaded['PAD.html'])
```

```python
from bs4 import BeautifulSoup
pagina = BeautifulSoup(data)
```

#EXplore

```python
tabelas = pagina.find_all('table', {'class':'participantes'})   # <table> : Defines a table
print(len(tabelas))
```

```python
tabela = tabelas[1]
```

```python
alunos = tabela.find_all('td', {'valign':'top'})    # <td> : Defines a cell in a table

print(len(alunos))
```

```python
nome, curso, matricula, usuario, email = [], [], [], [], []
i=0
for aluno in alunos:
    nome.append(((aluno.find('strong')).getText()).strip())
    dados = (aluno.find_all('em'))
    curso.append(dados[0].getText())
    matricula.append(dados[1].getText())
    usuario.append(dados[2].getText())
    email.append(dados[3].getText())
    i=i+1

print(i)

ne = i
```

```python
import pandas as pd
df = pd.DataFrame({'Nome': nome, 'Curso': curso, 'Matricula': matricula, 'Usuario':usuario, 'Email':email})
df.head()
```

```python
import unicodedata

def remove_acentos(text):
  return unicodedata.normalize('NFKD', text).encode('ASCII', 'ignore').decode('ASCII')

def acha_matricula(nome):
  for i in range(ne):
    if nome in df.Nome[i]:
      print()
      print(f'Nome: {df.Nome[i]:<40} \t Email: {df.Email[i]}')

nome_input = input('Digite o nome, por favor: ')
acha_matricula(remove_acentos(nome_input).upper()  )
```
